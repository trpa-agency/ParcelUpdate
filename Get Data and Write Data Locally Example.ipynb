# Download Test Points 
Example here: https://gis.stackexchange.com/questions/331863/copying-feature-service-to-local-geodatabase-using-arcpy-and-arcgis-api-for-pyth

```python

########################################################################################################################

GDB_Name = 'Test_20220118'# input("Name Geodatabase: ")

Folder_Input = 'C:\working_folder' # input("Provide Folder Path of Project: ")

# REST_URL_for_Feature = 'https://services6.arcgis.com/1KtlSd2mklZMBKaz/arcgis/rest/services/CA_Dspace_v2019/FeatureServer/0'#input("Provide : REST Item URL")

REST_URL_for_Feature = "https://services6.arcgis.com/1KtlSd2mklZMBKaz/arcgis/rest/services/ExPointLinePoly/FeatureServer/0"

Feature_Class_Name = 'Points'#input("Provide Name for New Feature Class: ")

########################################################################################################################

WorkspaceGDB = arcpy.CreateFileGDB_management(Folder_Input, '{}.gdb'.format(GDB_Name))

arcpy.env.workspace = r'{}'.format(WorkspaceGDB)

print("GDB Created")
########################################################################################################################

arcpy.CopyFeatures_management(REST_URL_for_Feature, Feature_Class_Name)

print("Features copied to geodatabase")
print("Process Finished")
```
